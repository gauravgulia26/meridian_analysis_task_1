### India Mart Pricing Table

 Steel Sheet Price Breakdown per kg in India
• 
📌 CR-3 Cold Rolled Steel

	• Price Range: ₹62 – ₹77 per kg
	• Context: Standard deep-drawing cold-rolled close-annealed (CRCA) steel used for general automotive panels and appliances.
• 
📌 IF (Interstitial-Free) Steel

	• Price Range: ₹60 – ₹72 per kg
	• Context: Ultra-low carbon steel with high formability for complex deep-drawn automotive body components.
• 
📌 HSLA 340 Steel

	• Price Range: ₹68 – ₹81 per kg
	• Context: High-Strength Low-Alloy steel providing ~340 MPa yield strength for structural reinforcement and chassis parts.
• 
📌 DP600 (Dual-Phase) Steel

	• Price Range: ₹72 – ₹82 per kg
	• Context: Advanced high-strength steel (AHSS) carrying an approximate 12–18% premium over standard CR grades.
• 
📌 DP780 (Dual-Phase) Steel

	• Price Range: ₹75 – ₹88 per kg
	• Context: Higher tensile dual-phase automotive grade used for crash-relevant structures and safety beams.

```python
df = df.merge(material_price_df, on="material_grade_clean", how="left")
```

```python
df["material_cost_low"] = df["unit_weight_kg"] * df["price_low_per_kg"]

df["material_cost_base"] = df["unit_weight_kg"] * df["price_base_per_kg"]

df["material_cost_high"] = df["unit_weight_kg"] * df["price_high_per_kg"]
```
