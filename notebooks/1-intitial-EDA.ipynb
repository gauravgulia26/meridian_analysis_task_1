```python
# --------------------------------------------------
# 1. Load data
# --------------------------------------------------

df = pd.read_csv("/home/gaurav/Documents/statxo/data/meridian_spend_data.csv")

# Fix known typo
df["category"] = df["category"].replace({"Farsteners": "Fasteners"})


# --------------------------------------------------
# 2. Existing categories
# --------------------------------------------------

choices = {category: category for category in df["category"].dropna().unique()}


# --------------------------------------------------
# 3. Create classifier
# --------------------------------------------------

classifier = TypeSafeClassifier(
    api_key="apikey_211639f0f2295086454db7ac587e7dce8148_91e82280b8bdcab71764600193ce459d09c4b84906ece7a9a7727e80b9bfa4e7"
)


# --------------------------------------------------
# 4. Classify one part
# --------------------------------------------------


def classify_part(description: str, current_category: str | None):

    response = classifier.invoke(
        {
            "state": (
                f"Part description: {description}\n"
                f"Current category: {current_category}\n\n"
                "Determine whether the current category accurately "
                "represents the part described. If the category is "
                "incorrect, identify the most appropriate category "
                "from the available categories."
            ),
            "questions": {
                "category_match": Noul(
                    instructions=(
                        "Does the current category accurately match "
                        "the part description?"
                    )
                ),
                "recommended_category": Choice(
                    instructions=(
                        "What is the most appropriate category for "
                        "this part based on its description?"
                    ),
                    criteria=choices,
                ),
            },
        }
    )

    return response
```

```python
row = range(df.shape[0])
```

```python
requests = []

for _, row in df.iterrows():
    requests.append(
        {
            "state": (
                f"Part description: {row['part_description']}\n"
                f"Current category: {row['category']}"
            ),
            "questions": {
                "category_match": Noul(
                    instructions=(
                        "Does the current category accurately match "
                        "the part description?"
                    )
                ),
                "recommended_category": Choice(
                    instructions=(
                        "What is the most appropriate category for "
                        "this part based on its description?"
                    ),
                    criteria=choices,
                ),
            },
        }
    )
```

```python
responses = classifier.batch(requests, config={"max_concurrency": 20})
```

```python
results = []

for row, response in zip(df.itertuples(index=False), responses):
    recommended = response.choices["recommended_category"]

    results.append(
        {
            "part_description": row.part_description,
            "previous_category": row.category,
            "category_match_probability": (response.nouls["category_match"].noul),
            "jev_category": recommended.choice,
            "jev_confidence": recommended.confidence,
            "category_probabilities": recommended.probabilities,
        }
    )

classification_df = pd.DataFrame(results)
```
